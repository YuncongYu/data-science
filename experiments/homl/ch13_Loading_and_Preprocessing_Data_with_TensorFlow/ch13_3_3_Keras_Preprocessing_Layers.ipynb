# Chapter 13.3.3 Keras Preprocessing Layers

##### Normalization Layer

```python
import numpy as np
import tensorflow as tf

data = np.arange(12).reshape(4, 3)

normalization = tf.keras.layers.Normalization()
normalization.adapt(data)
print(data)
print(normalization(data))
```

##### Discretization Layer

```python
data = np.arange(18).reshape(6, 3)

discretization = tf.keras.layers.Discretization(num_bins=3)
discretization.adapt(data)
discretization(data)
```

##### TextVectorization Layer

```python
data = [
    "The greatest glory lies in rising every time we fall.",
    "The way to get started is to quit talking and begin doing.",
    "Don't waste it living someone else's life.",
    "If life were predictable it would lose flavor",
    "The way we see the life",
]

text_vectorization = tf.keras.layers.TextVectorization()
text_vectorization.adapt(data)
text_vectorization(data)
```

```python
text_vectorization = tf.keras.layers.TextVectorization(output_mode="multi_hot")
text_vectorization.adapt(data)
text_vectorization(data)
```

```python
text_vectorization = tf.keras.layers.TextVectorization(output_mode="count")
text_vectorization.adapt(data)
text_vectorization(data)
```

```python
text_vectorization = tf.keras.layers.TextVectorization(output_mode="tf_idf")
text_vectorization.adapt(data)
text_vectorization(data)
```

##### How TensorFlow / Keras Calculates TF-IDF

```python
# Term counts
terms = ["the", "way", "we", "see", "lift"]

# Term frequency
tf = np.array([2, 1, 1, 1, 1])

# Number of documents
D = 5

# Number of documents containing the term
d = np.array([3, 2, 2, 1, 3])

# Inverse document frequency
idf = np.log(1 + D / (d + 1))

# tf-idf
tf * idf
```
