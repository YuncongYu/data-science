# Section 1: Using TensorFlow like NumPy

```python
import numpy as np
import tensorflow as tf
```

## 1.1 Tensors and Operations

Scaler

```python
tf.constant(42)
```

Matrix

```python
t = tf.constant([[1., 2, 3], [4, 5, 6]])
t
```

Shape and data type

```python
t.shape
```

```python
t.dtype
```

Indexing

```python
t[:, 1:]
```

```python
t[..., 1]
```

```python
t[..., 1, tf.newaxis]
```

```python
t[..., tf.newaxis, 1]
```

```python
t[tf.newaxis, ..., 1]
```

```python
t[..., 1, tf.newaxis, tf.newaxis]
```

Tensor operations

```python
t * t
```

```python
t_t = tf.transpose(t)
t_t
```

```python
t @ t_t
```

```python
m = np.arange(1, 7).reshape(2, 3)
m
```

```python
m @ m.T
```

```python
a = [[1, 2, 3], [4, 5, 6]]
a
```

```python
b = list(zip(*a))
b
```

```python
# Error unsupported operand type
# a @ b
```

Math operations

```python
tf.reduce_mean(t)
```

Keras' low-level API

```python
import tensorflow.keras.backend as K

K.square(K.transpose(t))
```

## 1.4 Variables

```python
v = tf.Variable([[1, 2, 3], [4, 5, 6]], dtype=float)
v
```

```python
w = 2 * v
w
```

```python
v.assign(2 * v)
```

```python
v[0, 1].assign(42)
```

```python
v[:, 2].assign([100, 200])
```

```python
v.scatter_nd_update(indices=[[0, 0], [1, 2]], updates=[0.1, 0.2])
```
