# Tensorflow Functions and Graphs

### Introduction

```python
import tensorflow as tf

tf_square = tf.function(lambda x: x**2)
tf_square
```

```python
tf_square(2.0)
```

```python
@tf.function
def tf_square(x):
    return x**2


tf_square
```

```python
tf_square(3.0)
```

```python
import numpy as np

print(np.shape(3))
print(np.shape([3]))
```

```python
tf.autograph.to_code(tf_square.python_function)
```

```python
@tf.function
def random_num(x):
    return np.random.rand()


print(random_num(1))
print(random_num(2))
print(random_num(tf.constant(1)))
print(random_num(tf.constant(2)))
```

```python
@tf.function
def random_num(x):
    return tf.random.uniform([])


print(random_num(1))
print(random_num(2))
print(random_num(tf.constant(1)))
print(random_num(tf.constant(2)))
```

### TF Functions and Concrete Functions

```python
concrete_function = tf_square.get_concrete_function(tf.constant([2.0, 3]))
concrete_function
```
