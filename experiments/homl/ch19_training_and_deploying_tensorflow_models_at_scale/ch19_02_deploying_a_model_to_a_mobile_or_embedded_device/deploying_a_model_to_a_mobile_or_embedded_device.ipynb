# Chapter 19: Deploying a Model to Mobile or Embedded Device

```python
from pathlib import Path

import tensorflow as tf

path_model = Path() / "bert"
converter = tf.lite.TFLiteConverter.from_saved_model(str(path_model))
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_model = converter.convert()
with open("model_converted_optimized_for_size.tflite", "wb") as fp:
    fp.write(tflite_model)
```
