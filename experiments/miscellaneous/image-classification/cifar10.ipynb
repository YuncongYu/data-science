# Object Classification with CIFAR-10

### 1. CIFAR-10 Photo Classification Dataset
CIFAR - Canadian Institute For Advanced Research.

```python
from typing import Optional, Sequence

import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline
plt.style.use("dark_background")
```

##### 1.1 Load Dataset

```python
from tensorflow.keras.datasets import cifar10

(X_train, y_train), (X_test, y_test) = cifar10.load_data()
image_classes = np.asarray(
    [
        "airplane",
        "automobile",
        "bird",
        "cat",
        "deer",
        "dog",
        "frog",
        "horse",
        "ship",
        "truck",
    ]
)

print(f"x_train.shape = {X_train.shape}")
print(f"y_train.shape = {y_train.shape}")
print(f"x_test.shape = {X_test.shape}")
print(f"y_test.shape = {y_test.shape}")

print(f"Feature range: {X_train.min(), X_train.max()}")
```

##### 1.2 Plot Data

```python
def plot_images(
    images: Sequence[Sequence[Sequence[Sequence[float]]]],
    image_labels: Optional[Sequence[str]],
    n_cols: int = 5,
    subfig_edge_len: float = 2,
) -> None:
    # Prepare inputs
    if image_labels is None:
        image_labels = [None] * len(images)

    # Initialize figure
    n_rows = int(np.ceil(len(images) / n_cols))
    figsize = (subfig_edge_len * n_cols, subfig_edge_len * n_rows)
    fig, axs = plt.subplots(nrows=n_rows, ncols=n_cols, figsize=figsize)

    # Plot every image
    for ax, image, image_label in zip(axs.ravel(), images, image_labels):
        ax.imshow(image)
        ax.set_axis_off()
        if image_label:
            ax.set_title(image_label)

    # Plot setting
    plt.show()


plot_images(X_train[:10], image_classes[y_train[:10].ravel()])
```
