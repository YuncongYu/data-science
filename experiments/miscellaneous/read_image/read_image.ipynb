# Read Image

```python
import numpy as np

path_image = "labrador.JPEG"
```

## 1 PIL

```python
from PIL import Image

img = Image.open(path_image, "r")
type(img)
```

```python
img
```

```python
img.mode
```

```python
img.size
```

```python
img.format
```

```python
img.rotate(45)  # in deg
```

```python
img.resize([224, 224])
```

```python
path_image_new = "new_img.jpg"
img.save(path_image_new)
Image.open(path_image)
```

```python
import matplotlib.pyplot as plt

data = np.asarray(img)
print(data.shape)
plt.imshow(data);  # cmap="viridis"
```

## 2 ImageIO

```python
import imageio.v3 as iio

img = iio.imread(path_image)
type(img)
```

```python
plt.imshow(img);
```

```python
iio.imwrite(path_image_new, img)
img = iio.imread(path_image_new)
plt.imshow(img);
```

## 3 OpenCV

```python
import cv2

img = cv2.imread(path_image)
type(img)
```

## 4 Scikit-Image

```python
import skimage

img = skimage.io.imread(path_image)
type(img)
```

```python
skimage.io.imshow(img)
```

## 4 Matplotlib

```python
import matplotlib.image as mpimg
import matplotlib.pyplot as plt

img = mpimg.imread(path_image)
type(img)
```

```python
plt.imshow(img)
```

```python
import tensorflow as tf

img = tf.keras.preprocessing.image.load_img(path_image)
type(img)
```
